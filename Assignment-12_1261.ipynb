```python
# LAB ASSIGNMENT 12
# VALUE ITERATION ALGORITHM IN A DETERMINISTIC GRID WORLD

# Step 1: Define the MDP

gamma = 0.9
theta = 0.000001

states = ["S0", "S1", "G"]
actions = ["Up", "Down", "Left", "Right"]

terminal_state = "G"

normal_reward = -1
goal_reward = 100

# Step 2: Define deterministic transition rules

def get_transition(state, action):

    transitions = {
        "S0": {
            "Up": ("S0", normal_reward),
            "Down": ("S1", normal_reward),
            "Left": ("S0", normal_reward),
            "Right": ("S0", normal_reward)
        },

        "S1": {
            "Up": ("S0", normal_reward),
            "Down": ("S1", normal_reward),
            "Left": ("S1", normal_reward),
            "Right": ("G", goal_reward)
        }
    }

    return transitions[state][action]


# Step 3: Initialize state values to zero

V = {
    "S0": 0.0,
    "S1": 0.0,
    "G": 0.0
}

print("INITIAL STATE VALUES")
print("--------------------------------")
print("V(S0) =", V["S0"])
print("V(S1) =", V["S1"])
print("V(G)  =", V["G"])


# Step 4: Display state, action, reward and transition information

print("\nSTATE TRANSITION INFORMATION")
print("--------------------------------")

for state in ["S0", "S1"]:

    for action in actions:

        next_state, reward = get_transition(state, action)

        print(
            f"State: {state} | Action: {action} | "
            f"Next State: {next_state} | Reward: {reward}"
        )


# Step 5: Calculate action values

def calculate_q_values(state, V):

    q_values = {}

    for action in actions:

        next_state, reward = get_transition(
            state, action
        )

        q_value = reward + gamma * V[next_state]

        q_values[action] = q_value

    return q_values


# Step 6: Implement Value Iteration

iteration = 0

print("\nVALUE ITERATION PROCESS")
print("================================")

while True:

    delta = 0

    old_values = V.copy()

    # Update values for all non-terminal states

    for state in ["S0", "S1"]:

        q_values = calculate_q_values(state, old_values)

        best_value = max(q_values.values())

        delta = max(
            delta,
            abs(best_value - old_values[state])
        )

        V[state] = best_value

    iteration += 1

    # Display iteration-wise state values

    print(f"\nIteration {iteration}")

    print(f"V(S0) = {V['S0']:.6f}")
    print(f"V(S1) = {V['S1']:.6f}")
    print(f"Delta = {delta:.6f}")

    # Check convergence

    if delta < theta:
        print("\nConvergence reached!")
        break


# Step 7: Calculate final action values

print("\nFINAL ACTION VALUE TABLE")
print("================================")

final_q_values = {}

for state in ["S0", "S1"]:

    q_values = calculate_q_values(state, V)

    final_q_values[state] = q_values

    for action, value in q_values.items():

        print(
            f"Q({state}, {action}) = {value:.2f}"
        )


# Step 8: Find optimal policy

optimal_policy = {}

print("\nOPTIMAL POLICY")
print("================================")

for state in ["S0", "S1"]:

    q_values = final_q_values[state]

    best_action = max(
        q_values,
        key=q_values.get
    )

    optimal_policy[state] = best_action

    print(
        f"State: {state} | "
        f"Optimal Action: {best_action} | "
        f"Optimal Value: {V[state]:.2f}"
    )


# Step 9: Display final optimal state values

print("\nFINAL OPTIMAL STATE VALUES")
print("================================")

for state in states:

    print(
        f"V*({state}) = {V[state]:.2f}"
    )


# Step 10: Bellman Optimality Verification

print("\nBELLMAN OPTIMALITY VERIFICATION")
print("================================")

for state in ["S0", "S1"]:

    max_q = max(final_q_values[state].values())

    print(f"\nState: {state}")
    print(f"V*({state}) = {V[state]:.2f}")
    print(f"Maximum Q*({state}, a) = {max_q:.2f}")

    if abs(V[state] - max_q) < theta:
        print("Verification: Successful")
    else:
        print("Verification: Failed")


# Step 11: Display optimal path

print("\nOPTIMAL PATH")
print("================================")
print("S0 --> S1 --> G")

print("\nValue Iteration Completed Successfully!")
```
