```python
# LAB ASSIGNMENT 8
# POLICY ITERATION ALGORITHM IN A DETERMINISTIC GRID WORLD

# Step 1: Define the MDP

gamma = 0.9
theta = 0.000001

states = ["S0", "S1", "G"]
actions = ["Up", "Down", "Left", "Right"]

terminal_state = "G"

# Rewards
normal_reward = -1
goal_reward = 100

# Deterministic transition function
def get_transition(state, action):

    transitions = {
        "S0": {
            "Up": ("S0", normal_reward),
            "Down": ("S1", normal_reward),
            "Left": ("S0", normal_reward),
            "Right": ("S0", normal_reward)
        },

        "S1": {
            "Up": ("S0", normal_reward),
            "Down": ("S1", normal_reward),
            "Left": ("S1", normal_reward),
            "Right": ("G", goal_reward)
        }
    }

    return transitions[state][action]


# Step 2: Initialize policy and state values

policy = {
    "S0": "Up",
    "S1": "Up"
}

V = {
    "S0": 0.0,
    "S1": 0.0,
    "G": 0.0
}

print("INITIAL POLICY")
print("------------------------------")
print("S0:", policy["S0"])
print("S1:", policy["S1"])

# Display state transitions

print("\nSTATE TRANSITIONS")
print("------------------------------")

for state in ["S0", "S1"]:
    for action in actions:
        next_state, reward = get_transition(state, action)
        print(state, "--", action, "-->",
              next_state, "| Reward:", reward)


# Step 3: Policy Evaluation

def policy_evaluation(policy, V):

    iteration = 0

    while True:

        delta = 0
        old_values = V.copy()

        for state in ["S0", "S1"]:

            action = policy[state]

            next_state, reward = get_transition(
                state, action
            )

            new_value = reward + gamma * old_values[next_state]

            delta = max(
                delta,
                abs(new_value - old_values[state])
            )

            V[state] = new_value

        iteration += 1

        if delta < theta:
            break

    return V, iteration


# Step 4: Calculate action values

def calculate_q_values(state, V):

    q_values = {}

    for action in actions:

        next_state, reward = get_transition(
            state, action
        )

        q_value = reward + gamma * V[next_state]

        q_values[action] = q_value

    return q_values


# Step 5: Policy Improvement

def policy_improvement(policy, V):

    policy_stable = True

    for state in ["S0", "S1"]:

        old_action = policy[state]

        q_values = calculate_q_values(state, V)

        best_action = max(
            q_values,
            key=q_values.get
        )

        policy[state] = best_action

        if old_action != best_action:
            policy_stable = False

    return policy, policy_stable


# Step 6: Implement Policy Iteration

policy_iteration = 0

print("\nPOLICY ITERATION PROCESS")
print("==============================")

while True:

    policy_iteration += 1

    # Policy Evaluation
    V, eval_iterations = policy_evaluation(policy, V)

    print("\nIteration:", policy_iteration)
    print("Policy Evaluation Iterations:", eval_iterations)

    print("State Values:")
    print("V(S0) =", round(V["S0"], 6))
    print("V(S1) =", round(V["S1"], 6))

    # Calculate action values
    print("\nAction Values:")

    for state in ["S0", "S1"]:

        q_values = calculate_q_values(state, V)

        print("\n", state)

        for action, value in q_values.items():
            print(action, ":", round(value, 6))

    # Policy Improvement
    old_policy = policy.copy()

    policy, stable = policy_improvement(policy, V)

    print("\nUpdated Policy:")
    print("S0:", policy["S0"])
    print("S1:", policy["S1"])

    # Check Policy Stability
    if stable:
        print("\nPolicy is stable!")
        break
    else:
        print("\nPolicy changed. Repeating...")


# Step 7: Display Final Optimal Results

print("\n====================================")
print("FINAL OPTIMAL RESULTS")
print("====================================")

print("\nOptimal State Values:")
for state in states:
    print(state, ":", round(V[state], 2))

print("\nOptimal Policy:")
for state in ["S0", "S1"]:
    print(state, ":", policy[state])

print("\nFinal Action Value Table:")
print("------------------------------------")

for state in ["S0", "S1"]:

    q_values = calculate_q_values(state, V)

    for action, value in q_values.items():

        print(
            f"Q({state}, {action}) = {value:.2f}"
        )


# Step 8: Bellman Optimality Verification

print("\nBELLMAN OPTIMALITY VERIFICATION")
print("------------------------------------")

for state in ["S0", "S1"]:

    q_values = calculate_q_values(state, V)

    max_q = max(q_values.values())

    print(f"\nState: {state}")
    print(f"V*({state}) = {V[state]:.2f}")
    print(f"Maximum Q*({state}, a) = {max_q:.2f}")

    if abs(V[state] - max_q) < theta:
        print("Verification: Successful")
    else:
        print("Verification: Failed")


# Step 9: Display Optimal Path

print("\nOPTIMAL PATH")
print("------------------------------------")
print("S0 --> S1 --> G")
print("\nPolicy Iteration Completed Successfully!")
```
