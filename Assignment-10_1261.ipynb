```python
# Assignment 1: Bellman Optimality and Value Iteration
# Deterministic Grid World

# --------------------------------------------------
# 1. Define the MDP (States, Actions, Rewards)
# --------------------------------------------------

states = ['S0', 'S1', 'G']

actions = ['UP', 'DOWN', 'LEFT', 'RIGHT']

gamma = 0.9
threshold = 0.0001

# Grid layout:
# S0 | WALL
# S1 | G

wall = 'WALL'
goal = 'G'

# Rewards
normal_reward = -1
wall_reward = -1
goal_reward = 10

# Deterministic transition function
def get_next_state(state, action):

    # Terminal state
    if state == goal:
        return goal, 0

    # Possible transitions
    transitions = {
        'S0': {
            'UP': ('S0', normal_reward),
            'DOWN': ('S1', normal_reward),
            'LEFT': ('S0', normal_reward),
            'RIGHT': ('S0', wall_reward)
        },
        'S1': {
            'UP': ('S0', normal_reward),
            'DOWN': ('S1', normal_reward),
            'LEFT': ('S1', normal_reward),
            'RIGHT': ('G', goal_reward)
        }
    }

    return transitions[state][action]


# --------------------------------------------------
# 2. Display MDP Information
# --------------------------------------------------

print("=" * 60)
print("1. MDP INFORMATION")
print("=" * 60)

print("States:", states)
print("Actions:", actions)
print("Discount factor (gamma):", gamma)
print("Normal movement reward:", normal_reward)
print("Wall reward:", wall_reward)
print("Goal reward:", goal_reward)
print("Terminal state:", goal)

print("\nGrid World:")
print("+---------+---------+")
print("|   S0    |  WALL   |")
print("+---------+---------+")
print("|   S1    |    G    |")
print("+---------+---------+")


# --------------------------------------------------
# 3. Verify All State-Action Transitions
# --------------------------------------------------

print("\n" + "=" * 60)
print("2. TRANSITION VERIFICATION")
print("=" * 60)

for state in ['S0', 'S1']:
    for action in actions:
        next_state, reward = get_next_state(state, action)
        print(
            f"State: {state:2} | Action: {action:5} | "
            f"Next state: {next_state:2} | Reward: {reward}"
        )


# --------------------------------------------------
# 4. Initialize State Values
# --------------------------------------------------

V = {
    'S0': 0.0,
    'S1': 0.0,
    'G': 0.0
}

print("\n" + "=" * 60)
print("3. INITIAL STATE VALUES")
print("=" * 60)

print(V)


# --------------------------------------------------
# 5. Implement Value Iteration
# --------------------------------------------------

print("\n" + "=" * 60)
print("4. VALUE ITERATION")
print("=" * 60)

iteration = 0

while True:
    iteration += 1

    # Synchronous update: use previous iteration values
    old_V = V.copy()
    new_V = old_V.copy()

    for state in ['S0', 'S1']:

        q_values = []

        for action in actions:
            next_state, reward = get_next_state(state, action)

            q = reward + gamma * old_V[next_state]
            q_values.append(q)

        new_V[state] = max(q_values)

    # Goal is terminal
    new_V['G'] = 0.0

    # Calculate maximum change
    delta = max(
        abs(new_V[state] - old_V[state])
        for state in states
    )

    V = new_V

    print(
        f"Iteration {iteration:2}: "
        f"S0 = {V['S0']:8.4f}, "
        f"S1 = {V['S1']:8.4f}, "
        f"G = {V['G']:6.2f}, "
        f"Delta = {delta:.6f}"
    )

    if delta < threshold:
        break


# --------------------------------------------------
# 6. Calculate Final Q-values
# --------------------------------------------------

print("\n" + "=" * 60)
print("5. FINAL ACTION VALUES Q*(s, a)")
print("=" * 60)

Q = {}

for state in ['S0', 'S1']:
    Q[state] = {}

    for action in actions:
        next_state, reward = get_next_state(state, action)

        Q[state][action] = (
            reward + gamma * V[next_state]
        )

        print(
            f"Q({state}, {action}) = "
            f"{Q[state][action]:.4f}"
        )


# --------------------------------------------------
# 7. Extract the Optimal Policy
# --------------------------------------------------

print("\n" + "=" * 60)
print("6. OPTIMAL POLICY")
print("=" * 60)

policy = {}

for state in ['S0', 'S1']:

    best_value = max(Q[state].values())

    # Include every action in case of a tie
    best_actions = [
        action for action, value in Q[state].items()
        if abs(value - best_value) < 1e-9
    ]

    policy[state] = best_actions

    print(
        f"State {state}: Optimal action(s) = "
        f"{best_actions}, Q-value = {best_value:.4f}"
    )


# --------------------------------------------------
# 8. Verify Bellman Optimality Equation
# --------------------------------------------------

print("\n" + "=" * 60)
print("7. BELLMAN OPTIMALITY VERIFICATION")
print("=" * 60)

for state in ['S0', 'S1']:

    max_q = max(Q[state].values())

    print(
        f"State {state}: V* = {V[state]:.4f}, "
        f"max Q* = {max_q:.4f}, "
        f"Verified = {abs(V[state] - max_q) < 1e-9}"
    )


# --------------------------------------------------
# 9. Explain the Results
# --------------------------------------------------

print("\n" + "=" * 60)
print("8. FINAL RESULTS AND EXPLANATION")
print("=" * 60)

print("\nOptimal State Values:")
for state in states:
    print(f"V*({state}) = {V[state]:.4f}")

print("\nExplanation:")
print("1. The agent chooses actions with the highest Q-value.")
print("2. Moving into the wall keeps the agent in the same state.")
print("3. The goal gives a reward of +10 and is terminal.")
print("4. The discount factor gamma = 0.9 reduces future rewards.")
print("5. Value iteration stops when the maximum change is")
print("   smaller than the chosen threshold.")
print("6. The optimal policy gives the best action in each state.")

print("\nAssignment completed successfully!")
```
