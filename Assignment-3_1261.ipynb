```python
import random

# ============================================================
# Part A - Machine Initialization
# ============================================================

machines = ['A', 'B', 'C']

M = 3
N_play = 10
epsilon = 0.2

# ============================================================
# Part B - Initialize N, R and Q
# ============================================================

N = {machine: 0 for machine in machines}
R = {machine: 0 for machine in machines}
Q = {machine: 0.0 for machine in machines}

# Given reward patterns
reward_patterns = {
    'A': [3, 4, 2, 5],
    'B': [6, 7, 5, 8],
    'C': [4, 3, 5, 4]
}

# Keep track of which reward to use next for each machine
reward_index = {
    'A': 0,
    'B': 0,
    'C': 0
}

print("Initial Values")
print("N =", N)
print("R =", R)
print("Q =", Q)
print("-" * 60)


# ============================================================
# Part C - Modified Epsilon-Greedy Decision
# ============================================================

for play in range(1, N_play + 1):

    # Generate random number
    r = random.random()

    # Find highest Q-value
    max_Q = max(Q.values())

    # Find machines having highest Q
    highest_machines = [
        machine for machine in machines
        if Q[machine] == max_Q
    ]

    # Choose highest-Q machine
    highest_machine = random.choice(highest_machines)

    # --------------------------------------------------------
    # Exploration
    # --------------------------------------------------------

    if r < epsilon:

        # Exclude highest-Q machine
        other_machines = [
            machine for machine in machines
            if machine != highest_machine
        ]

        selected_machine = random.choice(other_machines)

        decision = "Exploration"

    # --------------------------------------------------------
    # Exploitation
    # --------------------------------------------------------

    else:

        selected_machine = highest_machine

        decision = "Exploitation"


    # ========================================================
    # Part D - Automatically Collect Reward
    # ========================================================

    # Get the next reward from the given pattern
    index = reward_index[selected_machine]

    reward = reward_patterns[selected_machine][index]

    # Move to next reward
    # % 4 makes the pattern repeat if needed
    reward_index[selected_machine] = (index + 1) % 4


    # ========================================================
    # Part E - Update N, R and Q
    # ========================================================

    # Increase participation count
    N[selected_machine] += 1

    # Add reward
    R[selected_machine] += reward

    # Incremental sample-average formula
    Q[selected_machine] = (
        Q[selected_machine]
        + (reward - Q[selected_machine]) / N[selected_machine]
    )


    # ========================================================
    # Display Current Play
    # ========================================================

    print(f"\nPlay {play}")
    print(f"Random number = {r:.4f}")
    print(f"Decision = {decision}")
    print(f"Selected Machine = {selected_machine}")
    print(f"Reward = {reward}")
    print(f"Updated N = {N[selected_machine]}")
    print(f"Updated R = {R[selected_machine]}")
    print(f"Updated Q = {Q[selected_machine]:.2f}")


# ============================================================
# Part F - Final Results
# ============================================================

print("\n" + "=" * 70)
print("FINAL RESULTS")
print("=" * 70)

print("\nMachine\tN (Count)\tR (Total Reward)\tQ (Estimated Reward)")
print("-" * 70)

for machine in machines:

    print(
        f"{machine}\t"
        f"{N[machine]}\t\t"
        f"{R[machine]:.2f}\t\t\t"
        f"{Q[machine]:.2f}"
    )


# ============================================================
# Total Reward
# ============================================================

total_reward = sum(R.values())

print("\nTotal Reward Received =", total_reward)


# ============================================================
# Best Machine
# ============================================================

best_machine = max(Q, key=Q.get)

print("Best Machine =", best_machine)
print("Highest Q-value =", round(Q[best_machine], 2))
```
