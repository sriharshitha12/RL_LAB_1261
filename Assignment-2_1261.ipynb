```python
import random

# Part A – Machine Initialization

machines = ['A', 'B', 'C']

# Number of plays
N_play = 10

# Exploration probability
epsilon = 0.2


# Part B – Initialize N, R and Q tables

# N = Number of times each machine is selected
N = {
    'A': 0,
    'B': 0,
    'C': 0
}

# R = Store rewards received from each machine
R = {
    'A': [],
    'B': [],
    'C': []
}

# Q = Estimated value of each machine
Q = {
    'A': 0,
    'B': 0,
    'C': 0
}

total_reward = 0


# Part C, D and E – ε-Greedy Decision,
# Reward Collection and Q-Value Update

for play in range(1, N_play + 1):

    print("\n---------------------------")
    print("Play", play)
    print("---------------------------")

    # Generate random number between 0 and 1
    r = random.random()

    # ε-Greedy decision
    if r < epsilon:

        # Exploration
        selected_machine = random.choice(machines)

        print("Exploration")
        print("Random number =", round(r, 2))

    else:

        # Exploitation
        selected_machine = max(Q, key=Q.get)

        print("Exploitation")
        print("Random number =", round(r, 2))

    print("Play with machine", selected_machine)

    # Part D – Get reward from user
    reward = float(input("Enter reward: "))

    # Store reward
    R[selected_machine].append(reward)

    # Update participation count
    N[selected_machine] += 1

    # Part E – Incremental Sample Average Update
    Q[selected_machine] = Q[selected_machine] + (
        reward - Q[selected_machine]
    ) / N[selected_machine]

    # Add reward to total reward
    total_reward += reward

    # Display updated values
    print("\nUpdated Values")

    print("N =", N)
    print("Q =", Q)


# Part F – Display Final Results

print("\n================================")
print("FINAL RESULTS")
print("================================")

for machine in machines:

    print("\nMachine", machine)
    print("Participation Count N =", N[machine])
    print("Rewards R =", R[machine])
    print("Q Value =", round(Q[machine], 2))


print("\n================================")
print("Total Reward after 10 plays =", total_reward)
print("================================")
```
