```python
import math

# ============================================================
# Part A - Machine Initialization
# ============================================================

machines = ['A', 'B', 'C']

M = 3
N_play = 10

# ============================================================
# Part B - Initialize N, R, UCB and Q
# ============================================================

# Number of times each machine is selected
N = {machine: 0 for machine in machines}

# Total reward received from each machine
R = {machine: 0 for machine in machines}

# Estimated reward of each machine
Q = {machine: 0.0 for machine in machines}

# UCB values
UCB = {machine: float('inf') for machine in machines}


# Given reward patterns
reward_patterns = {
    'A': [3, 4, 2, 5],
    'B': [6, 7, 5, 8],
    'C': [4, 3, 5, 4]
}

# Keep track of next reward for each machine
reward_index = {
    'A': 0,
    'B': 0,
    'C': 0
}

print("Initial Values")
print("N =", N)
print("R =", R)
print("UCB =", UCB)
print("Q =", Q)
print("-" * 70)


# ============================================================
# Part C - UCB Calculation and Machine Selection
# ============================================================

for play in range(1, N_play + 1):

    # --------------------------------------------------------
    # Calculate UCB
    # --------------------------------------------------------

    for machine in machines:

        if N[machine] == 0:
            UCB[machine] = float('inf')

        else:
            UCB[machine] = Q[machine] + math.sqrt(
                (2 * math.log(play)) / N[machine]
            )

    # Select machine having highest UCB
    selected_machine = max(UCB, key=UCB.get)


    # ========================================================
    # Part D - Reward Collection
    # ========================================================

    index = reward_index[selected_machine]

    reward = reward_patterns[selected_machine][index]

    # Move to next reward
    reward_index[selected_machine] = (index + 1) % 4


    # ========================================================
    # Part E - Q-Value Update
    # ========================================================

    # Update number of selections
    N[selected_machine] += 1

    # Update total reward
    R[selected_machine] += reward

    # Incremental sample-average formula
    Q[selected_machine] = (
        Q[selected_machine]
        + (reward - Q[selected_machine]) / N[selected_machine]
    )


    # ========================================================
    # Display Play Information
    # ========================================================

    print(f"\nPlay {play}")

    print("Selected Machine =", selected_machine)
    print("Reward =", reward)

    print("N =", N[selected_machine])
    print("R =", R[selected_machine])
    print("Q =", round(Q[selected_machine], 2))

    print("UCB Values:")

    for machine in machines:
        if UCB[machine] == float('inf'):
            print(f"  {machine} = Infinity")
        else:
            print(f"  {machine} = {UCB[machine]:.4f}")


# ============================================================
# Part F - Final Results
# ============================================================

print("\n" + "=" * 70)
print("FINAL RESULTS")
print("=" * 70)

print("\nMachine\tN (Count)\tR (Total Reward)\tUCB\t\tQ")

print("-" * 70)

# Recalculate final UCB values
for machine in machines:

    if N[machine] == 0:
        UCB[machine] = float('inf')

    else:
        UCB[machine] = Q[machine] + math.sqrt(
            (2 * math.log(N_play)) / N[machine]
        )

    if UCB[machine] == float('inf'):
        ucb_display = "Infinity"
    else:
        ucb_display = f"{UCB[machine]:.4f}"

    print(
        f"{machine}\t"
        f"{N[machine]}\t\t"
        f"{R[machine]:.2f}\t\t\t"
        f"{ucb_display}\t"
        f"{Q[machine]:.2f}"
    )


# ============================================================
# Total Reward
# ============================================================

total_reward = sum(R.values())

print("\nTotal Reward Received =", total_reward)

# Best machine according to Q
best_machine = max(Q, key=Q.get)

print("Best Machine =", best_machine)
print("Highest Q-value =", round(Q[best_machine], 2))
```
