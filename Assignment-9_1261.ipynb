```python
# ASSIGNMENT 1
# Bellman Expectation Equations - Algebraic Method

gamma = 0.9

# Step 1: Bellman equations
# V(S0) = -1 + 0.81 * V(S1)
# V(S1) = 89.9 + 0.09 * V(S0)

print("BELLMAN EXPECTATION EQUATIONS")
print("--------------------------------")
print("V(S0) = -1 + 0.81 * V(S1)")
print("V(S1) = 89.9 + 0.09 * V(S0)")

# Step 2: Solve simultaneous equations algebraically
#
# Substitute V(S1) into V(S0):
# V(S0) = -1 + 0.81 * (89.9 + 0.09 * V(S0))
#
# V(S0) = 71.819 + 0.0729 * V(S0)
#
# 0.9271 * V(S0) = 71.819

V_S0 = 71.819 / 0.9271
V_S1 = 89.9 + 0.09 * V_S0

print("\nINTERMEDIATE CALCULATIONS")
print("--------------------------------")
print(f"V(S0) = 71.819 / 0.9271")
print(f"V(S0) = {V_S0:.6f}")
print(f"V(S1) = 89.9 + 0.09 * V(S0)")
print(f"V(S1) = {V_S1:.6f}")

# Step 3: Calculate action values

Q_S0_Down = -1 + gamma * V_S1
Q_S0_Right = -1

Q_S1_Right = 0.9 * 100 + 0.1 * (-1 + gamma * V_S0)
Q_S1_Up = -1 + gamma * V_S0

# Step 4: Display results

print("\nFINAL RESULTS - ASSIGNMENT 1")
print("--------------------------------")
print(f"V(S0)       = {V_S0:.2f}")
print(f"V(S1)       = {V_S1:.2f}")
print(f"Q(S0, Down) = {Q_S0_Down:.2f}")
print(f"Q(S0, Right)= {Q_S0_Right:.2f}")
print(f"Q(S1, Right)= {Q_S1_Right:.2f}")
print(f"Q(S1, Up)   = {Q_S1_Up:.2f}")

# Step 5: Summary table

print("\nSUMMARY TABLE")
print("--------------------------------")
print("{:<18} {:>12}".format("State / Action", "Value"))
print("-" * 32)

results = [
    ("V(S0)", V_S0),
    ("V(S1)", V_S1),
    ("Q(S0, Down)", Q_S0_Down),
    ("Q(S0, Right)", Q_S0_Right),
    ("Q(S1, Right)", Q_S1_Right),
    ("Q(S1, Up)", Q_S1_Up)
]

for name, value in results:
    print("{:<18} {:>12.2f}".format(name, value))

# Step 6: Verify Bellman equations

print("\nVERIFICATION")
print("--------------------------------")

check_S0 = 0.9 * Q_S0_Down + 0.1 * Q_S0_Right
check_S1 = 0.9 * Q_S1_Right + 0.1 * Q_S1_Up

print(f"V(S0) = {V_S0:.6f}")
print(f"Expected V(S0) = {check_S0:.6f}")

print(f"V(S1) = {V_S1:.6f}")
print(f"Expected V(S1) = {check_S1:.6f}")

print("\nAssignment 1 completed successfully!")
```

```python
# ASSIGNMENT 2
# Iterative Policy Evaluation - Reinforcement Learning

# Step 1: Initialize parameters

gamma = 0.9
theta = 0.000001

V_S0 = 0.0
V_S1 = 0.0

iterations = 0

print("ITERATIVE POLICY EVALUATION")
print("--------------------------------")
print("Discount factor (gamma):", gamma)
print("Convergence threshold:", theta)

# Step 2: Perform iterative policy evaluation

while True:

    old_V_S0 = V_S0
    old_V_S1 = V_S1

    # Calculate new state values using old values

    # S0 -> Wall (0.1) or S1 (0.9)
    new_V_S0 = (
        0.1 * (-1 + gamma * 0) +
        0.9 * (-1 + gamma * old_V_S1)
    )

    # S1 -> Goal (0.9) or S0 (0.1)
    new_V_S1 = (
        0.9 * (100 + gamma * 0) +
        0.1 * (-1 + gamma * old_V_S0)
    )

    # Calculate maximum change

    delta = max(
        abs(new_V_S0 - old_V_S0),
        abs(new_V_S1 - old_V_S1)
    )

    # Update state values

    V_S0 = new_V_S0
    V_S1 = new_V_S1

    iterations += 1

    # Stop when convergence is reached

    if delta < theta:
        break

# Step 3: Display convergence results

print("\nCONVERGENCE RESULTS")
print("--------------------------------")
print("Number of iterations:", iterations)
print(f"V(S0) = {V_S0:.6f}")
print(f"V(S1) = {V_S1:.6f}")
print(f"Final delta = {delta:.10f}")

# Step 4: Calculate action values after convergence

# S0, Down: Wall (0.1), S1 (0.9)
Q_S0_Down = (
    0.1 * (-1 + gamma * 0) +
    0.9 * (-1 + gamma * V_S1)
)

# S0, Right: Wall (1.0)
Q_S0_Right = -1 + gamma * 0

# S1, Right: Goal (0.9), S0 (0.1)
Q_S1_Right = (
    0.9 * (100 + gamma * 0) +
    0.1 * (-1 + gamma * V_S0)
)

# S1, Up: S0 (1.0)
Q_S1_Up = -1 + gamma * V_S0

# Step 5: Display action values

print("\nACTION VALUES")
print("--------------------------------")
print(f"Q(S0, Down)  = {Q_S0_Down:.2f}")
print(f"Q(S0, Right) = {Q_S0_Right:.2f}")
print(f"Q(S1, Right) = {Q_S1_Right:.2f}")
print(f"Q(S1, Up)    = {Q_S1_Up:.2f}")

# Step 6: Policy consistency verification

print("\nPOLICY CONSISTENCY CHECK")
print("--------------------------------")

print(f"V(S0)       = {V_S0:.6f}")
print(f"Q(S0, Down) = {Q_S0_Down:.6f}")

print(f"\nV(S1)       = {V_S1:.6f}")
print(f"Q(S1, Right)= {Q_S1_Right:.6f}")

if (abs(V_S0 - Q_S0_Down) < theta and
    abs(V_S1 - Q_S1_Right) < theta):
    print("\nPolicy consistency verified successfully!")
else:
    print("\nPolicy consistency verification failed.")

# Step 7: Results table

print("\nFINAL RESULTS TABLE")
print("--------------------------------")
print("{:<20} {:>12}".format("State / Action", "Value"))
print("-" * 34)

results = [
    ("Number of iterations", iterations),
    ("V(S0)", V_S0),
    ("V(S1)", V_S1),
    ("Q(S0, Down)", Q_S0_Down),
    ("Q(S0, Right)", Q_S0_Right),
    ("Q(S1, Right)", Q_S1_Right),
    ("Q(S1, Up)", Q_S1_Up)
]

for name, value in results:
    if name == "Number of iterations":
        print("{:<20} {:>12}".format(name, value))
    else:
        print("{:<20} {:>12.2f}".format(name, value))

print("\nAssignment 2 completed successfully!")
```
