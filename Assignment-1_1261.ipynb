```python
n = int(input("Enter the number of machines: "))

machines = {}

for i in range(n):
    name = input(f"\nEnter machine {i + 1} name: ")

    rewards = list(map(int, input(
        f"Enter rewards for Machine {name} separated by space: "
    ).split()))

    machines[name] = rewards


q_values = {}

print("\n--- Estimated Q Values ---")

for machine, rewards in machines.items():

    q = sum(rewards) / len(rewards)
    q_values[machine] = q

    print(f"\nMachine {machine}")
    print("Rewards:", rewards)
    print(f"Q({machine}) = {q}")


# Greedy selection
selected_machine = max(q_values, key=q_values.get)

print("\nGreedy Selected Machine:", selected_machine)
```
