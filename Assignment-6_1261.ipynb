```python
import gymnasium as gym
import numpy as np


# ---------------------------------------------------------
# RANDOM AGENT FUNCTION
# ---------------------------------------------------------

def random_agent(environment_name, episodes=10):

    print("\n" + "=" * 60)
    print("Environment:", environment_name)
    print("=" * 60)

    # 1. Create the environment
    env = gym.make(environment_name)

    episode_rewards = []

    # Repeat for each episode
    for episode in range(1, episodes + 1):

        # 2 & 3. Reset environment and observe initial state
        state, info = env.reset()

        total_reward = 0
        done = False
        step = 0

        # Continue until episode terminates
        while not done:

            # 4 & 5. Identify action space and select random action
            action = env.action_space.sample()

            # 6 & 7. Execute action and observe next state and reward
            next_state, reward, terminated, truncated, info = env.step(action)

            # Add reward
            total_reward += reward
            step += 1

            # 8. Move to next state
            state = next_state

            # Check whether episode has terminated
            done = terminated or truncated

        # Store total reward of this episode
        episode_rewards.append(total_reward)

        # 10. Print total reward for each episode
        print(
            "Episode", episode,
            "| Steps:", step,
            "| Total Reward:", total_reward
        )

    # 9. Calculate average reward
    average_reward = np.mean(episode_rewards)

    print("-" * 60)
    print("Total Reward of all episodes:", sum(episode_rewards))
    print("Average Reward:", average_reward)
    print("-" * 60)

    env.close()

    return episode_rewards, average_reward


# ---------------------------------------------------------
# RUN RANDOM AGENT ON CLIFF WALKING
# ---------------------------------------------------------

cliff_rewards, cliff_average = random_agent(
    "CliffWalking-v1",
    episodes=10
)


# ---------------------------------------------------------
# RUN RANDOM AGENT ON TAXI
# ---------------------------------------------------------

taxi_rewards, taxi_average = random_agent(
    "Taxi-v4",
    episodes=10

)
```
